```python
#checking for outliers using only numerical features so without label
numeric_df = df.drop("label", axis=1)

import matplotlib.pyplot as plt

numeric_df.boxplot(figsize=(12,6))
plt.xticks(rotation=45)
plt.title("Detection of Outliers in Numerical Features")
plt.show()


```
